## Check Missing

```python
all_data.isnull().sum()
```

```python
missing_data_by_station = all_data.groupby('tele_station_oldcode')['rainfall_value'].apply(lambda x: x.isnull().sum())
total_data_by_station = all_data.groupby('tele_station_oldcode')['rainfall_value'].count()

missing_df = pd.DataFrame({
    'missing_count': missing_data_by_station,
    'total_count': total_data_by_station + missing_data_by_station # Include missing in total count
})
missing_df['missing_proportion'] = missing_df['missing_count'] / missing_df['total_count']

# Sort by missing proportion in descending order for easier review
missing_df = missing_df.sort_values(by='missing_proportion', ascending=False)

print("ข้อมูลสูญหาย (Missing Data) ในแต่ละสถานี:")
print(missing_df.head(10)) # Display top 10 stations with most missing data

print("\nสรุปข้อมูลสูญหายโดยรวม:")
total_missing = all_data['rainfall_value'].isnull().sum()
total_entries = len(all_data)
total_missing_proportion = total_missing / total_entries
print(f"จำนวนข้อมูลสูญหายทั้งหมด: {total_missing} จาก {total_entries} แถว")
print(f"สัดส่วนข้อมูลสูญหายทั้งหมด: {total_missing_proportion:.2%}")
```

### คัดสถานีที่มีข้อมูลสูญหาย >= 20% ออก

```python
# Identify stations to drop (missing proportion >= 20%)
stations_to_drop = missing_df[missing_df['missing_proportion'] >= 0.20].index.tolist()

# Filter all_data to keep only stations with missing proportion < 20%
all_data_cleaned = all_data[~all_data['tele_station_oldcode'].isin(stations_to_drop)].copy()

# Get remaining stations
remaining_stations = all_data_cleaned['tele_station_oldcode'].unique()

print(f"สถานีที่ถูกคัดออกเนื่องจากมีข้อมูลสูญหาย >= 20% ({len(stations_to_drop)} สถานี):\n{stations_to_drop}\n")
print(f"สถานีที่เหลืออยู่ ({len(remaining_stations)} สถานี):\n{remaining_stations[:10]}... (แสดงเพียง 10 สถานีแรก)\n")

print(f"จำนวนแถวใน all_data ก่อนคัดออก: {len(all_data)}")
print(f"จำนวนแถวใน all_data หลังคัดออก: {len(all_data_cleaned)}")

all_data = all_data_cleaned.copy()
```

### เติม missing ด้วยค่ากลาง

```python
median_rainfall = all_data['rainfall_value'].median()
all_data['rainfall_value'] = all_data['rainfall_value'].fillna(median_rainfall)

print(f"เติมข้อมูลสูญหายใน 'rainfall_value' ด้วยค่ามัธยฐาน: {median_rainfall:.2f}")
print("ตรวจสอบข้อมูลสูญหายอีกครั้ง:")
print(all_data.isnull().sum())
```

**อภิปราย : ทำไมถึงใช้ median แทน mean**

   การเลือกใช้ค่ามัธยฐาน (median) แทนค่าเฉลี่ย (mean) ในกรณีของข้อมูลปริมาณน้ำฝนนั้น มีเหตุผลหลักมาจากลักษณะการกระจายตัวของข้อมูลปริมาณน้ำฝน
   
  1. ข้อมูลปริมาณน้ำฝนมักมีการกระจายตัวแบบไม่สมมาตร (Skewed Distribution): โดยทั่วไปแล้ว ในแต่ละวันจะมีวันที่ไม่มีฝนตกเลย (ค่า 0) หรือมีฝนตกน้อยเป็นส่วนใหญ่ และมีวันที่ฝนตกหนักมากเป็นส่วนน้อย ทำให้ข้อมูลมีการกระจายตัวไปทางขวา (right-skewed)
   
  2. ความไวต่อค่าผิดปกติ (Outliers): ค่าเฉลี่ย (mean) จะอ่อนไหวต่อค่าผิดปกติหรือค่าที่สูงมากๆ ได้ง่าย หากมีวันที่ฝนตกหนักผิดปกติเพียงไม่กี่วัน ค่าเฉลี่ยจะถูกดึงให้สูงขึ้น ซึ่งอาจทำให้ค่าเฉลี่ยไม่สะท้อนปริมาณน้ำฝน 'โดยทั่วไป' ได้ดีนัก
   
  3. ความทนทานของมัธยฐาน (Robustness of Median): ค่ามัธยฐาน (median) คือค่ากลางของข้อมูลที่เรียงลำดับแล้ว ดังนั้นค่าที่สูงหรือต่ำมากๆ จะไม่ส่งผลกระทบต่อค่ามัธยฐานมากเท่าค่าเฉลี่ย การใช้มัธยฐานจึงช่วยให้ค่าที่นำมาเติมมีความสมจริงและเป็นตัวแทนของปริมาณน้ำฝนส่วนใหญ่ได้ดีกว่า โดยเฉพาะอย่างยิ่งเมื่อมัธยฐานเป็น 0.00 ซึ่งหมายความว่ากว่าครึ่งหนึ่งของข้อมูลปริมาณน้ำฝนคือ 0 มม. หรือน้อยกว่านั้น

# คำนวณผลรวมปริมาณน้ำฝนรายเดือน

```python
# Convert 'rainfall_datetime_calc' to datetime objects
#ต้องใช้ 'rainfall_datetime_calc' เพราะจะได้คำนวณวันเวลาได้ถูกต้อง
all_data['rainfall_datetime_calc'] = pd.to_datetime(all_data['rainfall_datetime_calc'])

# Extract year and month
all_data['year'] = all_data['rainfall_datetime_calc'].dt.year
all_data['month'] = all_data['rainfall_datetime_calc'].dt.month

# Group by station, year, and month, then sum 'rainfall_value'
monthly_rain = all_data.groupby(['tele_station_oldcode', 'year', 'month'])['rainfall_value'].sum().reset_index()

# Rename columns to match Y_monthly_rain_HII_operation
monthly_rain = monthly_rain.rename(columns={
    'tele_station_oldcode': 'Node',
    'rainfall_value': 'rain'
})

# Reorder columns to match Y_monthly_rain_HII_operation
monthly_rain = monthly_rain[['Node', 'rain', 'month', 'year']]

print("Monthly rainfall data formatted as requested:")
print(monthly_rain.head())
print("\nInfo of the new DataFrame:")
monthly_rain.info()
```
